```python
df["Order_Date"]=pd.to_datetime(df["Order_Date"])
```

```python
df.info()
```

```python
df["Year"] = df["Order_Date"].dt.year
print(df["Year"])
```

```python
df["Month"]=df["Order_Date"].dt.month_name()
print(df["Month"])
```

```python
#Calculate Revenue
df["Revenue"]=df["Quantity"]*df["Unit_Price"]
print(df["Revenue"])
```

```python
df["Cost"] = pd.to_numeric(df["Cost"].astype(str).str.replace(',', ''), errors='coerce')
df["Revenue"] = pd.to_numeric(df["Revenue"].astype(str).str.replace(',', ''), errors='coerce')

df["Cost"] = df["Cost"].fillna(0)
df["Revenue"] = df["Revenue"].fillna(0)

df["Profit"] = df["Revenue"] - df["Cost"]  
```

```python
#Calculate Profit Margin

df["Profit_Margin"]=df["Profit"]/df["Revenue"]*100
```

```python
#Product Analysis

Product_Sales=(
    df.groupby("Product_Name")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)
print(Product_Sales)
```

```python
#Region Analysis
Region_Sales=(
               df.groupby("Region")["Revenue"]
               .sum()  
               .sort_values(ascending=False)
)

print(Region_Sales)
```

```python
#Customer Analysis

Customerwise_Sales=(
    df.groupby("Customer_Name")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)
print(Customerwise_Sales)
                
```

```python
#Category Analysis

Category_Sales=(
    df.groupby("Category")
    .agg(
        Revenue=("Revenue","sum"),
        Profit=("Profit","sum"),
        Quantity=("Quantity","sum")
        )
        .sort_values("Revenue",
                     ascending=False)
)
print(Category_Sales)
```

```python
#Monthly Sales

Monthly_Sales=(
    df.groupby("Month")["Revenue"]
    .sum()
)

print(Monthly_Sales)
    
```

```python
#Find Best Product
best_product = (
    df.groupby("Product_Name")["Revenue"]
    .sum()
    .idxmax()
)

print(best_product)
    
```

```python
#Find Highest Revenue Customer
best_customer = (
    df.groupby("Customer_Name")["Revenue"]
    .sum()
    .idxmax()
)

print(best_customer)

best_customer = (
    df.groupby("Customer_Name")["Revenue"]
    .sum()
    .idxmax()
)

print(best_customer)
```

```python
#Find Most Profitable Product
Best_Profit_product = (
    df.groupby("Product_Name")["Profit"]
    .sum()
    .idxmax()
)

print(Best_Profit_product)
```

```python
#Calculate average revenue

Average_Revenue=np.mean(df["Revenue"])
print(Average_Revenue)
```

```python
# Revenue Visualization
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(
        Product_Sales.index,
        Product_Sales.values
)
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
           
```

```python
#Monthly Trend

Monthly_Sales=(
df.groupby(
    df["Order_Date"].dt.to_period("M")
)["Revenue"]
.sum()
)

plt.figure(figsize=(10,5))
plt.plot(
          Monthly_Sales.index.astype(str),
          Monthly_Sales.values,
          marker="o"
)

plt.title("Monthly_Revenue_Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Profit Analysis
profit_by_category = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(profit_by_category)
```

```python
#High-Value Orders(high-value orders as revenue greater than ₹50,000).

high_value_orders = df[
    df["Revenue"] > 50000
]

print(high_value_orders)
```

```python
#Customer Segmentation


# Step 1: Convert Revenue column to numeric
df["Revenue"] = pd.to_numeric(
    df["Revenue"],
    errors="coerce"
)

# Step 2: Calculate total revenue for each customer
customer_revenue = (
    df.groupby("Customer_Name")["Revenue"]
    .sum()
)

# Step 3: Define customer segmentation function
def segment_customer(revenue):

    if revenue >= 50000:
        return "High Value"

    elif revenue >= 20000:
        return "Medium Value"

    else:
        return "Low Value"

# Step 4: Apply the function to each customer's revenue
customer_segment = customer_revenue.apply(segment_customer)

# Step 5: Display the result
print(customer_segment)
```

```python
#Final Python Analysis
# Step 1: Convert numeric columns to numeric data types
numeric_columns = [
    "Revenue",
    "Cost",
    "Profit",
    "Quantity"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Step 2: Calculate the final analysis summary
summary = {
    "Total Revenue": df["Revenue"].sum(),
    "Total Cost": df["Cost"].sum(),
    "Total Profit": df["Profit"].sum(),
    "Average Revenue": df["Revenue"].mean(),
    "Average Profit": df["Profit"].mean(),
    "Total Quantity": df["Quantity"].sum(),
    "Total Orders": df["Order_ID"].nunique(),
    "Total Customers": df["Customer_ID"].nunique()
}

# Step 3: Print the results
for key, value in summary.items():
    print(f"{key}: {value:,.2f}")
```
